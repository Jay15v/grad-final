```python
# =========================
# Role 1 Baseline Decomposer (Modified Full Code)
# - Fixes the "how they differ" fragment bug
# - Keeps decomposition structural only
# - Adds stronger atomic splitting rules
# - Keeps constraints + quality scoring
# =========================

import re
from dataclasses import dataclass, asdict
from typing import List, Dict, Optional, Tuple


# ---------- Data ----------
@dataclass
class Step:
    id: int
    text: str
    source_span: Optional[str] = None


ALLOWED_VERBS = (
    "Define", "Compare", "Identify", "Outline", "Explain", "List", "Provide", "Summarize",
    "State", "Verify", "Extract", "Clarify", "Distinguish"
)

# ---------- Utilities ----------
def _norm(s: str) -> str:
    return re.sub(r"\s+", " ", s.strip())

def starts_with_allowed_verb(step: str) -> bool:
    s = step.strip()
    return any(s.startswith(v + " ") or s == v for v in ALLOWED_VERBS)

def extract_constraints(prompt: str) -> Dict[str, object]:
    """
    Extract non-verifiable presentation constraints as metadata.
    Keeps steps clean for verification/RAG.
    """
    p = prompt.lower()
    constraints: Dict[str, object] = {}

    # Language
    if re.search(r"\bin arabic\b|\barabic\b|بالعربي|باللغة العربية", p):
        constraints["language"] = "Arabic"
    elif re.search(r"\bin english\b|\benglish\b|بالانجليزي|باللغة الانجليزية", p):
        constraints["language"] = "English"

    # Style / audience
    if re.search(r"\b(simple|simpler|easy)\b|بسيط|بسهولة", p):
        constraints["style"] = "simple"
    if re.search(r"\b(beginner|for beginners)\b|مبتدئ", p):
        constraints["audience"] = "beginner"

    # Format
    if re.search(r"\btable\b|\btabular\b|جدول", p):
        constraints["format"] = "table"
    if re.search(r"\bbullets?\b|نقاط", p):
        constraints["format"] = constraints.get("format", "bullets")

    # Citations requirement
    if re.search(r"\bwith (citations|references)\b|\breferences\b|\bcitations\b|مراجع", p):
        constraints["citations_required"] = True

    return constraints

def split_clauses(prompt: str) -> List[str]:
    p = _norm(prompt)
    p = re.sub(r"[;\n]+", ". ", p)
    p = re.sub(r"[?!]+", ". ", p)
    p = re.sub(r"\s*\.\s*", ". ", p).strip()

    sentences = [x.strip() for x in p.split(". ") if x.strip()]

    # NOTE: intentionally NOT splitting on plain "and" to avoid breaking phrases.
    connectors = [r"\band then\b", r"\balso\b", r"\bin addition\b", r"\bthen\b", r"\bplus\b"]
    out: List[str] = []
    for s in sentences:
        chunks = [s]
        for c in connectors:
            new_chunks: List[str] = []
            for ch in chunks:
                parts = [x.strip() for x in re.split(c, ch, flags=re.IGNORECASE) if x.strip()]
                new_chunks.extend(parts)
            chunks = new_chunks
        out.extend(chunks)

    return [_norm(x) for x in out if _norm(x)]

def atomic_split(step: str) -> List[str]:
    """
    Safer atomic splitting:
    - Never leaves fragments without a verb.
    - Only splits if both resulting parts look like tasks.
    - Special handling for "Define X and Y".
    """
    s = _norm(step)

    # If it's "Define X and Y" -> make two Define steps
    if s.lower().startswith("define "):
        tail = s[len("Define "):]
        if re.search(r"\b and \b", tail.lower()):
            parts = [p.strip(" ,") for p in re.split(r"\b and \b", tail, flags=re.IGNORECASE) if p.strip(" ,")]
            if len(parts) >= 2:
                return [f"Define {parts[0]}"] + [f"Define {p}" for p in parts[1:]]
        return [s]

    # For other steps: split ONLY if both sides start with allowed verbs
    if re.search(r"\b and \b|\b or \b", s.lower()):
        parts = [p.strip() for p in re.split(r"\b and \b|\b or \b", s, flags=re.IGNORECASE) if p.strip()]
        if len(parts) >= 2 and all(starts_with_allowed_verb(p) for p in parts):
            return parts

    return [s]


# ---------- Clause Expansion (templates) ----------
def expand_clause(clause: str) -> List[str]:
    c = _norm(clause)
    lc = c.lower()

    # Comparison
    if re.search(r"\bcompare\b|\bcontrast\b|\bvs\b|\bversus\b|\bdifference\b", lc):
        raw = re.sub(r"^(compare|contrast)\s+", "", c, flags=re.IGNORECASE).strip()
        entities = [e.strip(" ,") for e in re.split(r"\bvs\b|\bversus\b|\band\b", raw, flags=re.IGNORECASE) if e.strip(" ,")]

        if len(entities) >= 2:
            a, b = entities[0], entities[1]
            steps = [f"Define {a}", f"Define {b}", f"Compare {a} and {b} across key aspects"]
            if re.search(r"\bexample(s)?\b", lc):
                steps += [f"Provide examples for {a}", f"Provide examples for {b}"]
            return steps

        steps = ["Identify the items to be compared", "Define each item", "Compare the items across key aspects"]
        if re.search(r"\bexample(s)?\b", lc):
            steps.append("Provide examples for each item")
        return steps

    # Definition
    if re.search(r"\bdefine\b|\bwhat is\b|\bmeaning of\b|\bdefinition\b", lc):
        m = re.search(r"\bdefine\s+(.+)$", c, flags=re.IGNORECASE) or re.search(r"\bwhat is\s+(.+)$", c, flags=re.IGNORECASE)
        if m:
            term = m.group(1).strip(" ?.")
            return [f"Define {term}"]
        return ["Identify the term to define", "Define the term"]

    # Why (causal) — ensure last step does NOT create a split fragment
    if re.search(r"\bwhy\b", lc):
        m = re.search(r"\bwhy\s+(.+)$", c, flags=re.IGNORECASE)
        phenomenon = m.group(1).strip(" ?.") if m else c
        return [
            f"Identify the phenomenon in question: {phenomenon}",
            "Identify the primary causal/mechanistic explanation category relevant to the phenomenon",
            "State the key conditions/assumptions required for the explanation",
            "Explain the causal chain linking mechanism to the observed outcome",
            "List plausible alternative explanations and distinguish them"
        ]

    # How-to (procedure)
    if re.search(r"\bhow to\b", lc):
        m = re.search(r"\bhow to\s+(.+)$", c, flags=re.IGNORECASE)
        goal = m.group(1).strip(" ?.") if m else c
        return [
            f"Identify goal: {goal}",
            f"Outline the ordered steps required for {goal}",
            f"Explain the purpose of each step for {goal}",
            f"Identify common failure points or edge cases for {goal}"
        ]

    # Explain/describe
    if re.search(r"\bexplain\b|\bdescribe\b|\bhow does\b", lc):
        return [
            f"Identify the concept/topic to explain in: '{c}'",
            "Break the explanation into atomic sub-points",
            "Define any required terms before using them",
            "Ensure each sub-point is supported by evidence"
        ]

    # List
    if re.search(r"\blist\b|\benumerate\b|\bgive me\b|\bprovide\b|\bshow\b", lc):
        steps = [f"Identify items requested in: '{c}'", "List the requested items in an ordered way"]
        if re.search(r"\bexplain\b|\bdetail(s)?\b|\bwith details\b", lc):
            steps.append("Explain each listed item briefly")
        return steps

    return [f"Clarify the request component: '{c}'"]


# ---------- Decompose ----------
def decompose_prompt_baseline(prompt: str) -> Dict[str, object]:
    prompt = _norm(prompt)
    if not prompt:
        return {"steps": [], "constraints": {}}

    constraints = extract_constraints(prompt)
    clauses = split_clauses(prompt)

    steps_text: List[Tuple[str, str]] = []  # (step_text, source_span)
    for clause in clauses:
        for st in expand_clause(clause):
            for piece in atomic_split(st):
                piece = _norm(piece)
                if piece:
                    steps_text.append((piece, clause))

    # Deduplicate preserving order
    seen = set()
    final_steps: List[Step] = []
    sid = 1
    for st, span in steps_text:
        key = st.lower()
        if key in seen:
            continue
        seen.add(key)
        final_steps.append(Step(id=sid, text=st, source_span=span))
        sid += 1

    return {"steps": [asdict(s) for s in final_steps], "constraints": constraints}


def quality_score(prompt: str, steps: List[Dict[str, object]]) -> Dict[str, float]:
    p_len = len(prompt.split())
    n = len(steps)
    if n == 0:
        return {"score": 0.0, "atomicity": 0.0, "verb_rate": 0.0, "redundancy": 0.0}

    def is_true_multi_task(text: str) -> bool:
        """
        Penalize only if 'and/or' connects two independent obligations.
        Do NOT penalize common single-task patterns like:
          - "List X and distinguish them"
          - "Identify X and validate Y"  (this is borderline, but we'll treat as multi-task if both sides are verb-start)
        """
        t = text.strip()

        # If we can split into two parts and BOTH parts start with a verb -> multi-task
        if re.search(r"\b and \b|\b or \b", t.lower()):
            parts = [p.strip() for p in re.split(r"\b and \b|\b or \b", t, flags=re.IGNORECASE) if p.strip()]
            if len(parts) >= 2:
                # If the step is "List ... and distinguish ..." keep it atomic (single obligation)
                if parts[0].startswith("List ") and parts[1].startswith("distinguish"):
                    return False

                # If both sides are verb-start tasks, it's truly multi-task
                if all(starts_with_allowed_verb(p) for p in parts):
                    return True

        return False

    # Atomicity: penalize only true multi-task steps
    non_atomic = sum(1 for s in steps if is_true_multi_task(s["text"]))
    atomicity = 1.0 - (non_atomic / n)

    # Verb start rate
    verb_ok = sum(1 for s in steps if starts_with_allowed_verb(s["text"]))
    verb_rate = verb_ok / n

    # Step count sanity
    expected_min = 2 if p_len < 12 else 3
    expected_max = 10 if p_len > 30 else 8
    count_ok = 1.0 if (expected_min <= n <= expected_max) else 0.6

    redundancy = 1.0
    score = 0.40 * atomicity + 0.35 * verb_rate + 0.25 * count_ok
    return {"score": float(score), "atomicity": float(atomicity), "verb_rate": float(verb_rate), "redundancy": float(redundancy)}



# ---------- Interactive test ----------
def run_interactive_baseline():
    while True:
        prompt = input("\nEnter a prompt (or 'exit'): ").strip()
        if prompt.lower() in {"exit", "quit"}:
            break
        out = decompose_prompt_baseline(prompt)
        steps = out["steps"]
        print("\nSteps:")
        for s in steps:
            print(f"{s['id']}. {s['text']}")
        print("\nConstraints:", out["constraints"])
        print("Quality:", quality_score(prompt, steps))

run_interactive_baseline()
```
