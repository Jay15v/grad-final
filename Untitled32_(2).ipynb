````python
# =========================
# Role 1 Baseline Decomposer (Modified Full Code)
# - Fixes the "how they differ" fragment bug
# - Keeps decomposition structural only
# - Adds stronger atomic splitting rules
# - Keeps constraints + quality scoring
# =========================

import re
from dataclasses import dataclass, asdict
from typing import List, Dict, Optional, Tuple


# ---------- Data ----------
@dataclass
class Step:
    id: int
    text: str
    source_span: Optional[str] = None


ALLOWED_VERBS = (
    "Define", "Compare", "Identify", "Outline", "Explain", "List", "Provide", "Summarize",
    "State", "Verify", "Extract", "Clarify", "Distinguish"
)

# ---------- Utilities ----------
def _norm(s: str) -> str:
    return re.sub(r"\s+", " ", s.strip())

def starts_with_allowed_verb(step: str) -> bool:
    s = step.strip()
    return any(s.startswith(v + " ") or s == v for v in ALLOWED_VERBS)

def extract_constraints(prompt: str) -> Dict[str, object]:
    """
    Extract non-verifiable presentation constraints as metadata.
    Keeps steps clean for verification/RAG.
    """
    p = prompt.lower()
    constraints: Dict[str, object] = {}

    # Language
    if re.search(r"\bin arabic\b|\barabic\b|بالعربي|باللغة العربية", p):
        constraints["language"] = "Arabic"
    elif re.search(r"\bin english\b|\benglish\b|بالانجليزي|باللغة الانجليزية", p):
        constraints["language"] = "English"

    # Style / audience
    if re.search(r"\b(simple|simpler|easy)\b|بسيط|بسهولة", p):
        constraints["style"] = "simple"
    if re.search(r"\b(beginner|for beginners)\b|مبتدئ", p):
        constraints["audience"] = "beginner"

    # Format
    if re.search(r"\btable\b|\btabular\b|جدول", p):
        constraints["format"] = "table"
    if re.search(r"\bbullets?\b|نقاط", p):
        constraints["format"] = constraints.get("format", "bullets")

    # Citations requirement
    if re.search(r"\bwith (citations|references)\b|\breferences\b|\bcitations\b|مراجع", p):
        constraints["citations_required"] = True

    return constraints

def split_clauses(prompt: str) -> List[str]:
    p = _norm(prompt)
    p = re.sub(r"[;\n]+", ". ", p)
    p = re.sub(r"[?!]+", ". ", p)
    p = re.sub(r"\s*\.\s*", ". ", p).strip()

    sentences = [x.strip() for x in p.split(". ") if x.strip()]

    # NOTE: intentionally NOT splitting on plain "and" to avoid breaking phrases.
    connectors = [r"\band then\b", r"\balso\b", r"\bin addition\b", r"\bthen\b", r"\bplus\b"]
    out: List[str] = []
    for s in sentences:
        chunks = [s]
        for c in connectors:
            new_chunks: List[str] = []
            for ch in chunks:
                parts = [x.strip() for x in re.split(c, ch, flags=re.IGNORECASE) if x.strip()]
                new_chunks.extend(parts)
            chunks = new_chunks
        out.extend(chunks)

    return [_norm(x) for x in out if _norm(x)]

def atomic_split(step: str) -> List[str]:
    """
    Safer atomic splitting:
    - Never leaves fragments without a verb.
    - Only splits if both resulting parts look like tasks.
    - Special handling for "Define X and Y".
    """
    s = _norm(step)

    # If it's "Define X and Y" -> make two Define steps
    if s.lower().startswith("define "):
        tail = s[len("Define "):]
        if re.search(r"\b and \b", tail.lower()):
            parts = [p.strip(" ,") for p in re.split(r"\b and \b", tail, flags=re.IGNORECASE) if p.strip(" ,")]
            if len(parts) >= 2:
                return [f"Define {parts[0]}"] + [f"Define {p}" for p in parts[1:]]
        return [s]

    # For other steps: split ONLY if both sides start with allowed verbs
    if re.search(r"\b and \b|\b or \b", s.lower()):
        parts = [p.strip() for p in re.split(r"\b and \b|\b or \b", s, flags=re.IGNORECASE) if p.strip()]
        if len(parts) >= 2 and all(starts_with_allowed_verb(p) for p in parts):
            return parts

    return [s]


# ---------- Clause Expansion (templates) ----------
def expand_clause(clause: str) -> List[str]:
    c = _norm(clause)
    lc = c.lower()

    # Comparison
    if re.search(r"\bcompare\b|\bcontrast\b|\bvs\b|\bversus\b|\bdifference\b", lc):
        raw = re.sub(r"^(compare|contrast)\s+", "", c, flags=re.IGNORECASE).strip()
        entities = [e.strip(" ,") for e in re.split(r"\bvs\b|\bversus\b|\band\b", raw, flags=re.IGNORECASE) if e.strip(" ,")]

        if len(entities) >= 2:
            a, b = entities[0], entities[1]
            steps = [f"Define {a}", f"Define {b}", f"Compare {a} and {b} across key aspects"]
            if re.search(r"\bexample(s)?\b", lc):
                steps += [f"Provide examples for {a}", f"Provide examples for {b}"]
            return steps

        steps = ["Identify the items to be compared", "Define each item", "Compare the items across key aspects"]
        if re.search(r"\bexample(s)?\b", lc):
            steps.append("Provide examples for each item")
        return steps

    # Definition
    if re.search(r"\bdefine\b|\bwhat is\b|\bmeaning of\b|\bdefinition\b", lc):
        m = re.search(r"\bdefine\s+(.+)$", c, flags=re.IGNORECASE) or re.search(r"\bwhat is\s+(.+)$", c, flags=re.IGNORECASE)
        if m:
            term = m.group(1).strip(" ?.")
            return [f"Define {term}"]
        return ["Identify the term to define", "Define the term"]

    # Why (causal) — ensure last step does NOT create a split fragment
    if re.search(r"\bwhy\b", lc):
        m = re.search(r"\bwhy\s+(.+)$", c, flags=re.IGNORECASE)
        phenomenon = m.group(1).strip(" ?.") if m else c
        return [
            f"Identify the phenomenon in question: {phenomenon}",
            "Identify the primary causal/mechanistic explanation category relevant to the phenomenon",
            "State the key conditions/assumptions required for the explanation",
            "Explain the causal chain linking mechanism to the observed outcome",
            "List plausible alternative explanations and distinguish them"
        ]

    # How-to (procedure)
    if re.search(r"\bhow to\b", lc):
        m = re.search(r"\bhow to\s+(.+)$", c, flags=re.IGNORECASE)
        goal = m.group(1).strip(" ?.") if m else c
        return [
            f"Identify goal: {goal}",
            f"Outline the ordered steps required for {goal}",
            f"Explain the purpose of each step for {goal}",
            f"Identify common failure points or edge cases for {goal}"
        ]

    # Explain/describe
    if re.search(r"\bexplain\b|\bdescribe\b|\bhow does\b", lc):
        return [
            f"Identify the concept/topic to explain in: '{c}'",
            "Break the explanation into atomic sub-points",
            "Define any required terms before using them",
            "Ensure each sub-point is supported by evidence"
        ]

    # List
    if re.search(r"\blist\b|\benumerate\b|\bgive me\b|\bprovide\b|\bshow\b", lc):
        steps = [f"Identify items requested in: '{c}'", "List the requested items in an ordered way"]
        if re.search(r"\bexplain\b|\bdetail(s)?\b|\bwith details\b", lc):
            steps.append("Explain each listed item briefly")
        return steps

    return [f"Clarify the request component: '{c}'"]


# ---------- Decompose ----------
def decompose_prompt_baseline(prompt: str) -> Dict[str, object]:
    prompt = _norm(prompt)
    if not prompt:
        return {"steps": [], "constraints": {}}

    constraints = extract_constraints(prompt)
    clauses = split_clauses(prompt)

    steps_text: List[Tuple[str, str]] = []  # (step_text, source_span)
    for clause in clauses:
        for st in expand_clause(clause):
            for piece in atomic_split(st):
                piece = _norm(piece)
                if piece:
                    steps_text.append((piece, clause))

    # Deduplicate preserving order
    seen = set()
    final_steps: List[Step] = []
    sid = 1
    for st, span in steps_text:
        key = st.lower()
        if key in seen:
            continue
        seen.add(key)
        final_steps.append(Step(id=sid, text=st, source_span=span))
        sid += 1

    return {"steps": [asdict(s) for s in final_steps], "constraints": constraints}


def quality_score(prompt: str, steps: List[Dict[str, object]]) -> Dict[str, float]:
    p_len = len(prompt.split())
    n = len(steps)
    if n == 0:
        return {"score": 0.0, "atomicity": 0.0, "verb_rate": 0.0, "redundancy": 0.0}

    def is_true_multi_task(text: str) -> bool:
        """
        Penalize only if 'and/or' connects two independent obligations.
        Do NOT penalize common single-task patterns like:
          - "List X and distinguish them"
          - "Identify X and validate Y"  (this is borderline, but we'll treat as multi-task if both sides are verb-start)
        """
        t = text.strip()

        # If we can split into two parts and BOTH parts start with a verb -> multi-task
        if re.search(r"\b and \b|\b or \b", t.lower()):
            parts = [p.strip() for p in re.split(r"\b and \b|\b or \b", t, flags=re.IGNORECASE) if p.strip()]
            if len(parts) >= 2:
                # If the step is "List ... and distinguish ..." keep it atomic (single obligation)
                if parts[0].startswith("List ") and parts[1].startswith("distinguish"):
                    return False

                # If both sides are verb-start tasks, it's truly multi-task
                if all(starts_with_allowed_verb(p) for p in parts):
                    return True

        return False

    # Atomicity: penalize only true multi-task steps
    non_atomic = sum(1 for s in steps if is_true_multi_task(s["text"]))
    atomicity = 1.0 - (non_atomic / n)

    # Verb start rate
    verb_ok = sum(1 for s in steps if starts_with_allowed_verb(s["text"]))
    verb_rate = verb_ok / n

    # Step count sanity
    expected_min = 2 if p_len < 12 else 3
    expected_max = 10 if p_len > 30 else 8
    count_ok = 1.0 if (expected_min <= n <= expected_max) else 0.6

    redundancy = 1.0
    score = 0.40 * atomicity + 0.35 * verb_rate + 0.25 * count_ok
    return {"score": float(score), "atomicity": float(atomicity), "verb_rate": float(verb_rate), "redundancy": float(redundancy)}



# ---------- Interactive test ----------
def run_interactive_baseline():
    while True:
        prompt = input("\nEnter a prompt (or 'exit'): ").strip()
        if prompt.lower() in {"exit", "quit"}:
            break
        out = decompose_prompt_baseline(prompt)
        steps = out["steps"]
        print("\nSteps:")
        for s in steps:
            print(f"{s['id']}. {s['text']}")
        print("\nConstraints:", out["constraints"])
        print("Quality:", quality_score(prompt, steps))

run_interactive_baseline()
````

````python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "microsoft/Phi-3-mini-4k-instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)
````

````python
SYSTEM_RULES = """You are a structuring engine.
Task: Convert the user prompt into an ordered list of atomic reasoning-check steps.

Rules:
- DO NOT answer the question.
- DO NOT add facts, explanations, or examples.
- ONLY produce structural obligations.
- Output ONLY valid JSON (no markdown, no backticks, no extra text).
- JSON must have exactly these keys: "steps" and "constraints".
- "steps" must be a list of objects with keys: "id", "text".
- "constraints" must be an object (dictionary).
"""
````

````python
# 3) Robust JSON extractor (brace balancing)
import json

def extract_first_json_object(text: str):
    start = text.find("{")
    if start == -1:
        raise ValueError("No '{' found")

    depth = 0
    in_str = False
    escape = False

    for i in range(start, len(text)):
        ch = text[i]

        if in_str:
            if escape:
                escape = False
            elif ch == "\\":
                escape = True
            elif ch == '"':
                in_str = False
        else:
            if ch == '"':
                in_str = True
            elif ch == "{":
                depth += 1
            elif ch == "}":
                depth -= 1
                if depth == 0:
                    candidate = text[start:i+1]
                    return json.loads(candidate)

    raise ValueError("JSON appears truncated (no matching closing brace)")
````

````python
import re, json, torch

def local_llm_segment_once(question: str, max_new_tokens: int = 220):
    messages = [
        {"role": "system", "content": SYSTEM_RULES},
        {"role": "user", "content": question}
    ]
    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            eos_token_id=tokenizer.eos_token_id
        )

    # decode ONLY generated tokens
    gen_ids = output[0, inputs["input_ids"].shape[1]:]
    text = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

    # strip markdown fences if any
    text_clean = re.sub(r"```json|```", "", text, flags=re.IGNORECASE).strip()

    parsed = extract_first_json_object(text_clean)
    return parsed, text  # parsed dict + raw text


def local_llm_segment(question: str, retries: int = 3):
    """
    Robust wrapper:
    - retry with increasing max_new_tokens
    - if still failing, do a repair pass that forces JSON-only output
    """
    token_budgets = [220, 320, 450]  # increase for truncation cases

    # 1) Normal attempts
    for attempt in range(min(retries, len(token_budgets))):
        try:
            parsed, raw = local_llm_segment_once(question, max_new_tokens=token_budgets[attempt])
            # minimal validation
            if isinstance(parsed.get("steps"), list) and len(parsed["steps"]) > 0 and isinstance(parsed.get("constraints"), dict):
                return parsed
            return {"steps": [], "constraints": {}, "error": "bad_schema", "raw": raw[:2000]}
        except Exception as e:
            last_err = str(e)
            last_raw = locals().get("raw", "")

    # 2) Repair attempt: ask model to output ONLY JSON again
    REPAIR_SYSTEM = """You output ONLY valid JSON. No markdown, no extra text.
Return exactly: {"steps":[{"id":1,"text":"..."}], "constraints":{}} with real steps."""
    repair_question = f"""Rewrite as JSON only.

QUESTION:
{question}
"""

    try:
        messages = [
            {"role": "system", "content": REPAIR_SYSTEM},
            {"role": "user", "content": repair_question},
        ]
        prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

        with torch.no_grad():
            output = model.generate(
                **inputs,
                max_new_tokens=450,
                do_sample=False,
                eos_token_id=tokenizer.eos_token_id
            )

        gen_ids = output[0, inputs["input_ids"].shape[1]:]
        text = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()
        text_clean = re.sub(r"```json|```", "", text, flags=re.IGNORECASE).strip()

        parsed = extract_first_json_object(text_clean)
        if isinstance(parsed.get("steps"), list) and len(parsed["steps"]) > 0 and isinstance(parsed.get("constraints"), dict):
            return parsed

        return {"steps": [], "constraints": {}, "error": "repair_bad_schema", "raw": text[:2000], "error_detail": last_err}

    except Exception as e:
        return {"steps": [], "constraints": {}, "error": "json_error", "error_detail": last_err, "raw": last_raw[:2000]}
````

````python
# =========================
# LLM segmentation for the SAME first 50 questions (df_50)
# Assumes:
#   - df_50 already created from df.head(50) and has columns ["row_idx","question","long_answers"]
#   - local_llm_segment(question) is already working (JSON-only, no json_error)
# =========================

import json
from tqdm import tqdm

def llm_segment_row(q: str):
    return local_llm_segment(str(q))

# 1) Run LLM decomposition
tqdm.pandas()
df_50["llm_decomp"] = df_50["question"].progress_apply(llm_segment_row)

# 2) Convenience columns
df_50["llm_steps"] = df_50["llm_decomp"].apply(lambda x: x.get("steps", []) if isinstance(x, dict) else [])
df_50["llm_constraints"] = df_50["llm_decomp"].apply(lambda x: x.get("constraints", {}) if isinstance(x, dict) else {})

# 3) Validate quickly (count errors / empty steps)
bad = []
for i, r in df_50.iterrows():
    out = r["llm_decomp"]
    err = out.get("error") if isinstance(out, dict) else "not_a_dict"
    steps = out.get("steps") if isinstance(out, dict) else None

    if err:
        bad.append((int(r["row_idx"]), "error", err))
    elif not isinstance(steps, list) or len(steps) == 0:
        bad.append((int(r["row_idx"]), "empty_steps", None))

print("Total:", len(df_50), "| Bad:", len(bad))
if bad:
    print("First 10 issues:", bad[:10])
else:
    print("✅ All 50 have LLM steps with no errors.")

# 4) Sanity print: first 3
for i in range(3):
    print("\n==============================")
    print("row_idx:", df_50.loc[i, "row_idx"])
    print("Q:", df_50.loc[i, "question"])
    print("Steps:")
    for s in df_50.loc[i, "llm_steps"]:
        # s is dict: {"id":..., "text":...}
        print(f"  {s.get('id')}. {s.get('text')}")
    print("Constraints:", df_50.loc[i, "llm_constraints"])

# 5) Save LLM results to Drive as JSONL
OUT_JSONL_LLM = "/content/drive/MyDrive/llm_seg_first_50.jsonl"
with open(OUT_JSONL_LLM, "w", encoding="utf-8") as f:
    for _, row in df_50.iterrows():
        rec = {
            "row_idx": int(row["row_idx"]),
            "question": row["question"],
            "long_answers": row.get("long_answers", None),
            "llm": row["llm_decomp"]
        }
        f.write(json.dumps(rec, ensure_ascii=False) + "\n")

print("Saved:", OUT_JSONL_LLM)
````

````python
# =========================
# Check LLM segmentation health for df_50
# - counts errors
# - counts empty steps
# - shows problematic rows + question text
# =========================

import pandas as pd

assert "llm_decomp" in df_50.columns, "Run the LLM segmentation first (df_50['llm_decomp'])."

issues = []

for _, row in df_50.iterrows():
    row_idx = int(row["row_idx"])
    q = row["question"]
    out = row["llm_decomp"]

    if not isinstance(out, dict):
        issues.append({"row_idx": row_idx, "issue": "not_a_dict", "detail": str(type(out)), "question": q})
        continue

    if out.get("error"):
        issues.append({"row_idx": row_idx, "issue": "llm_error", "detail": out.get("error"), "question": q})
        continue

    steps = out.get("steps", None)
    if not isinstance(steps, list):
        issues.append({"row_idx": row_idx, "issue": "steps_not_list", "detail": str(type(steps)), "question": q})
        continue

    if len(steps) == 0:
        issues.append({"row_idx": row_idx, "issue": "empty_steps", "detail": "no steps returned", "question": q})
        continue

    # optional: verify each step has id/text
    for s in steps:
        if not isinstance(s, dict) or "id" not in s or "text" not in s:
            issues.append({"row_idx": row_idx, "issue": "bad_step_format", "detail": str(s)[:200], "question": q})
            break

issues_df = pd.DataFrame(issues)

print("Total rows:", len(df_50))
print("Issues found:", len(issues_df))

if len(issues_df) == 0:
    print("✅ All 50 questions have valid LLM segmentation (no errors, non-empty steps).")
else:
    print("❌ Issues breakdown:")
    print(issues_df["issue"].value_counts())

    # Show all problematic rows
    display(issues_df.sort_values(["issue", "row_idx"]).reset_index(drop=True))

    # If you want to print full raw output for a specific bad row_idx:
    # bad_id = issues_df.iloc[0]["row_idx"]
    # print(df_50.loc[df_50["row_idx"]==bad_id, "llm_decomp"].values[0])
````

````python
# =========================
# 3-way answering experiment on FIRST 50
# Case A: use MANUAL segmentation as plan (CoT-like)
# Case B: use LLM segmentation as plan (CoT-like)
# Case C: NO segmentation provided (model uses its own internal reasoning)
# Reference: long_answers
# Metrics: BLEU + BERTScore
# =========================

# Assumes df_50 already exists and contains:
# - question, long_answers
# - manual_steps (list of dicts with {"id","text"})
# - llm_steps (list of dicts with {"id","text"})
# And Phi-3 model/tokenizer already loaded as: model, tokenizer

import torch
import sacrebleu
from bert_score import score as bertscore
from tqdm import tqdm

# -------------------------
# 1) Helpers
# -------------------------
def steps_to_plan(steps):
    if not steps:
        return ""
    if isinstance(steps, list) and len(steps) > 0 and isinstance(steps[0], dict):
        return "\n".join([f"{s.get('id', i+1)}. {s.get('text','')}".strip() for i, s in enumerate(steps)])
    if isinstance(steps, list):
        return "\n".join([f"{i+1}. {s}" for i, s in enumerate(steps)])
    return ""

ANSWER_SYSTEM = """You answer using ONLY the provided CONTEXT.
If the context does not contain the answer, reply exactly: NOT_IN_CONTEXT
Return ONLY the final answer text. No reasoning, no steps, no extra text.
""".strip()

def build_prompt(question: str, context: str, plan: str = None):
    if plan is None:
        # Case C: no segmentation provided
        return f"""{ANSWER_SYSTEM}

CONTEXT:
{context}

QUESTION:
{question}

FINAL ANSWER:
""".strip()

    # Case A/B: provide plan (segmentation) as the model's "own plan"
    return f"""{ANSWER_SYSTEM}

CONTEXT:
{context}

QUESTION:
{question}

PLAN (follow these steps internally):
{plan}

FINAL ANSWER:
""".strip()

def local_generate_answer(prompt: str, max_new_tokens: int = 128):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            eos_token_id=tokenizer.eos_token_id
        )

    # decode only generated tokens
    gen_ids = output[0, inputs["input_ids"].shape[1]:]
    ans = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

    # keep first line only (prevents accidental extra output)
    return ans.splitlines()[0].strip()

# -------------------------
# 2) Build plans for cases A/B
# -------------------------
df_50["manual_plan"] = df_50["manual_steps"].apply(steps_to_plan)
df_50["llm_plan"] = df_50["llm_steps"].apply(steps_to_plan)

# Reference = long_answers (as requested)
df_50["reference"] = df_50["long_answers"].fillna("").astype(str)

# -------------------------
# 3) Generate answers for 3 cases
# -------------------------
ans_manual = []
ans_llmseg = []
ans_noseg = []

for _, row in tqdm(df_50.iterrows(), total=len(df_50), desc="Answering 3 cases"):
    q = str(row["question"])
    ctx = str(row["reference"])

    # Case A: manual segmentation plan
    p_manual = build_prompt(q, ctx, plan=row["manual_plan"])
    ans_manual.append(local_generate_answer(p_manual, max_new_tokens=128))

    # Case B: llm segmentation plan
    p_llmseg = build_prompt(q, ctx, plan=row["llm_plan"])
    ans_llmseg.append(local_generate_answer(p_llmseg, max_new_tokens=128))

    # Case C: no segmentation
    p_noseg = build_prompt(q, ctx, plan=None)
    ans_noseg.append(local_generate_answer(p_noseg, max_new_tokens=128))

df_50["answer_manual_seg"] = ans_manual
df_50["answer_llm_seg"] = ans_llmseg
df_50["answer_no_seg"] = ans_noseg

# Quick view
df_50[["question", "answer_manual_seg", "answer_llm_seg", "answer_no_seg"]].head(3)
````

````python
# =========================
# 4) Evaluate BLEU + BERTScore against long_answers
# =========================

refs = df_50["reference"].tolist()

hyps_manual = df_50["answer_manual_seg"].tolist()
hyps_llmseg = df_50["answer_llm_seg"].tolist()
hyps_noseg  = df_50["answer_no_seg"].tolist()

# BLEU
bleu_manual = sacrebleu.corpus_bleu(hyps_manual, [refs]).score
bleu_llmseg = sacrebleu.corpus_bleu(hyps_llmseg, [refs]).score
bleu_noseg  = sacrebleu.corpus_bleu(hyps_noseg,  [refs]).score

print("BLEU (manual seg):", bleu_manual)
print("BLEU (llm seg)   :", bleu_llmseg)
print("BLEU (no seg)    :", bleu_noseg)

# BERTScore (semantic similarity)
P_m, R_m, F1_m = bertscore(hyps_manual, refs, lang="en", model_type="distilbert-base-uncased", verbose=False)
P_l, R_l, F1_l = bertscore(hyps_llmseg, refs, lang="en", model_type="distilbert-base-uncased", verbose=False)
P_n, R_n, F1_n = bertscore(hyps_noseg,  refs, lang="en", model_type="distilbert-base-uncased", verbose=False)

print("BERTScore-F1 (manual seg):", float(F1_m.mean()))
print("BERTScore-F1 (llm seg)   :", float(F1_l.mean()))
print("BERTScore-F1 (no seg)    :", float(F1_n.mean()))
````

````python
# =========================
# 5) Optional: per-question comparison table (BERTScore F1)
# =========================

df_50["bert_f1_manual"] = F1_m.cpu().numpy()
df_50["bert_f1_llmseg"] = F1_l.cpu().numpy()
df_50["bert_f1_noseg"]  = F1_n.cpu().numpy()

df_50["best_case"] = df_50[["bert_f1_manual","bert_f1_llmseg","bert_f1_noseg"]].idxmax(axis=1)

df_50[["row_idx","question","bert_f1_manual","bert_f1_llmseg","bert_f1_noseg","best_case"]].head(10)
````
