```python
NLI_MODEL_NAME = "facebook/bart-large-mnli"

tokenizer_nli = AutoTokenizer.from_pretrained(NLI_MODEL_NAME)
model_nli = AutoModelForSequenceClassification.from_pretrained(NLI_MODEL_NAME)

device = "cuda" if torch.cuda.is_available() else "cpu"
model_nli = model_nli.to(device)
model_nli.eval()

# For BART MNLI the label order is usually: 0=contradiction, 1=neutral, 2=entailment
print("NLI loaded ✅", NLI_MODEL_NAME, "on", device)
```

```python
@torch.no_grad()
def nli_probs(premise: str, hypothesis: str, max_length: int = 512):
    """
    Returns dict with probabilities for entailment/neutral/contradiction.
    premise = evidence chunk
    hypothesis = claim
    """
    inputs = tokenizer_nli(
        premise,
        hypothesis,
        truncation=True,
        max_length=max_length,
        return_tensors="pt"
    ).to(device)

    logits = model_nli(**inputs).logits[0]
    probs = torch.softmax(logits, dim=-1).detach().cpu().numpy()

    # Standard MNLI order (most models): [contradiction, neutral, entailment]
    return {
        "contradiction": float(probs[0]),
        "neutral": float(probs[1]),
        "entailment": float(probs[2]),
    }
```

```python
def hybrid_gate(claim: str, candidates: list, sim_min=0.30, cov_min=0.12, keep_top=4):
    """
    Keep only evidence chunks that look relevant:
    - similarity from FAISS
    - coverage from your keywords overlap
    """
    gated = []
    for e in candidates:
        sim = float(e.get("similarity", 0.0))
        cov = float(e.get("coverage", coverage_score(claim, e.get("text",""))))
        e = {**e, "coverage": cov, "similarity": sim}
        if sim >= sim_min and cov >= cov_min:
            gated.append(e)

    # sort by similarity * coverage to keep best "relevant" chunks
    gated.sort(key=lambda x: (x["similarity"] * (0.5 + x["coverage"])), reverse=True)
    return gated[:keep_top]


def verify_step_hybrid(claim: str, top_evidence: list,
                       sim_min=0.30, cov_min=0.12,
                       entail_supported=0.70,
                       entail_uncertain=0.40,
                       contradiction_block=0.50):
    """
    Hybrid verification:
    1) Gate evidence by similarity+coverage
    2) Run NLI on remaining evidence
    3) Decide Supported/Uncertain/Unsupported
    """
    gated = hybrid_gate(claim, top_evidence, sim_min=sim_min, cov_min=cov_min, keep_top=4)

    if not gated:
        return {
            "label": "Unsupported",
            "confidence": 0.0,
            "reason": "No relevant evidence after gating",
            "best_evidence": None,
            "nli_details": []
        }

    nli_details = []
    max_entail = 0.0
    best = None
    contradiction_flag = False
    max_contra = 0.0

    for e in gated:
        ev_text = e.get("text", "")
        probs = nli_probs(ev_text, claim)
        e_out = {**e, "nli": probs}
        nli_details.append(e_out)

        max_entail = max(max_entail, probs["entailment"])
        if best is None or probs["entailment"] > best["nli"]["entailment"]:
            best = e_out

        max_contra = max(max_contra, probs["contradiction"])
        if probs["contradiction"] >= contradiction_block:
            contradiction_flag = True

    # Decision rules (simple + strong)
    if contradiction_flag:
        label = "Unsupported"
        reason = f"Contradiction detected (max contradiction={max_contra:.2f})"
        confidence = 1.0 - max_entail
    elif max_entail >= entail_supported:
        label = "Supported"
        reason = f"Entailed by evidence (max entailment={max_entail:.2f})"
        confidence = max_entail
    elif max_entail >= entail_uncertain:
        label = "Uncertain"
        reason = f"Some support but not strong (max entailment={max_entail:.2f})"
        confidence = max_entail
    else:
        label = "Unsupported"
        reason = f"Not supported by evidence (max entailment={max_entail:.2f})"
        confidence = max_entail

    return {
        "label": label,
        "confidence": float(confidence),
        "reason": reason,
        "best_evidence": best,
        "nli_details": nli_details
    }
```

```python
def build_template_response(role3_out: list):
    supported = [r for r in role3_out if r["label"] == "Supported"]
    uncertain  = [r for r in role3_out if r["label"] == "Uncertain"]
    unsupported = [r for r in role3_out if r["label"] == "Unsupported"]

    # Global decision
    if unsupported:
        mode = "REFUSE_OR_PARTIAL"
    elif uncertain:
        mode = "CAUTIOUS"
    else:
        mode = "FULL"

    lines = []
    lines.append(f"OUTPUT MODE: {mode}\n")

    # Verified answer = ONLY supported steps
    lines.append("=== Verified Answer (Supported) ===")
    if not supported:
        lines.append("(No supported steps to answer safely.)")
    else:
        for r in supported:
            be = r.get("best_evidence") or {}
            src = be.get("source","")
            url = be.get("url","")
            title = be.get("title","")
            # evidence-only summary (no free generation)
            lines.append(f"- Step {r['step']}: {r['claim']}")
            if url or title or src:
                lines.append(f"  Source: {src} | {title} | {url}")

    # Uncertain section
    if uncertain:
        lines.append("\n=== Uncertain (Needs stronger evidence) ===")
        for r in uncertain:
            lines.append(f"- Step {r['step']}: {r['claim']}")
            lines.append(f"  Why: {r['reason']}")

    # Unsupported section
    if unsupported:
        lines.append("\n=== Unsupported (Cannot be answered with current evidence) ===")
        for r in unsupported:
            lines.append(f"- Step {r['step']}: {r['claim']}")
            lines.append(f"  Why: {r['reason']}")

    return "\n".join(lines)


final_response = build_template_response(role3_out)
print(final_response)
```
